```python
import pandas as pd 
df=pd.read_csv("employee_project_dataset_fixed.csv")
print(df.isnull().sum())#isnull() — checks which cells are empty or missing#sum() — counts how many missing cells each column has.
```

```python
# axis 1 is row 
#First, isnull() checks every cell: MISSING VALUE MEANS TRUE if value is there false 
#any-checks whether at least one value is True.
df[df.isnull().any(axis=1)] #displays the rows that contain at least one missing value. 
```

```python
# so this is fillna fills the median value in missing 
# df["salary"] selects the specified column
# 
df["salary"]=df["salary"].fillna(df["salary"].median())
print(df["salary"].isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
# ex:it,hr,it,hr,it so it so mean takes it like 
# it 3 so first one is index 0 if we dont mention [0] it wont come properly it returns like dtype: object
# hr 2
df["department"]=df["department"].fillna(df["department"].mode()[0])
print(df["department"].isnull().sum())
```

```python
# Why mode? Chennai, Mumbai, Hyderabad and other city names are categories, not number
df["city"] = df["city"].fillna(df["city"].mode()[0])
print(df["city"].isnull().sum())
```

```python
df["age"]=df["age"].fillna(df["age"].median())
print(df["age"].isnull().sum())
```

```python
df["experience_years"]=df["experience_years"].fillna(df["experience_years"].median())
print(df["experience_years"].isnull().sum())
```

```python
df["performance_score"]=df["performance_score"].fillna(df["performance_score"].median())
print(df["performance_score"].isnull().sum())
```

```python
df["monthly_sales"]=df["monthly_sales"].fillna(df["monthly_sales"].median())
print(df["monthly_sales"].isnull().sum())
```

```python
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
print(df["gender"].isnull().sum())
```

```python
df["contract_type"] = df["contract_type"].fillna(df["contract_type"].mode()[0])
print(df["contract_type"].isnull().sum())
```

```python
# pd.to_datetime() converts the column into a proper date format.
# 2026-10-26 data type date(datetime) type ->treat it as a separate data type in Pandas.
# dont use median for date type use mode 
df["joining_date"] = pd.to_datetime(df["joining_date"])
df["joining_date"] = df["joining_date"].fillna(df["joining_date"].mode()[0])
print(df["joining_date"].isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
df.dtypes
```

```python
# We filled the missing values earlier. Missing values often cause numeric columns to become floats
df[["age","experience_years"]].head(10)
```

```python
# astype(int) changes the data type from decimal numbers (float) to whole numbers (int)
df["age"] = df["age"].astype(int)
df["experience_years"] = df["experience_years"].astype(int)

print(df[["age", "experience_years"]].dtypes)
```

```python
# This counts the duplicate rows
print(df.duplicated().sum())
```

```python
# drop_duplicates()removes duplicate rows.
df = df.drop_duplicates()

print(df.duplicated().sum())
```

```python
# select_dtypes(include="object") to check text columns that Pandas recognizes as having the object data type, often used for text.
print(df.select_dtypes(include="object").columns) #to check which columns has text and to check inconsistent capitalisation exists using unique
```

```python
# to check inconsistent capitalisation exists
print(df["department"].unique())
```

```python
# In Pandas, .str lets us apply text operations to all the values in a column
# .strip() — Remove extra spaces
# .title() — Capitalize each word
df["department"] = df["department"].str.strip().str.title()
df["department"]
```

```python
df["city"] = df["city"].str.strip().str.title()
df["city"]
```

```python
df["gender"] = df["gender"].str.strip().str.title()
df["gender"]
```

```python
df["contract_type"] = df["contract_type"].str.strip().str.title()
df["contract_type"]
```

```python
df["name"] = df["name"].str.strip().str.title()
df["name"]
```

```python
# this shows the salary count, average, standard deviation, minimum, maximum, and quartiles.
df["salary"].describe()
```

```python
# IQR stands for Interquartile Range.IQR = Q3 − Q1    third quartile 75%-87250.000000 first quartile25%-64500.000000
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

IQR = Q3 - Q1


lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

outliers = df[(df["salary"] < lower_limit) | (df["salary"] > upper_limit)]

print("Potential salary outliers:")
print(outliers[["name", "salary"]])
print("Number of potential outliers:", len(outliers))

```

```python
df["salary"] = df["salary"].clip(lower=lower_limit, upper=upper_limit)

print(df["salary"].describe())
```

```python
before_summary = pd.DataFrame({
    "Metric": ["Rows", "Missing Values", "Duplicates"],
    "Before Cleaning": [65, "Check original data", 5],
    "After Cleaning": [len(df), df.isnull().sum().sum(), df.duplicated().sum()]
})

print(before_summary)
```

```python
print(df.select_dtypes(include="number").agg(
    ["mean", "median", "min", "max", "std"]
))
```

```python
department_summary = df.groupby("department").agg(
    employee_count=("name", "count"),
    average_salary=("salary", "mean"),
    average_age=("age", "mean"),
    average_experience=("experience_years", "mean")
)

print(department_summary)
```

```python
city_summary = df.groupby("city").agg(
    employee_count=("name", "count"),
    average_salary=("salary", "mean"),
    average_experience=("experience_years", "mean")
)

print(city_summary)
```

```python
correlation_matrix = df.select_dtypes(include="number").corr()
print(correlation_matrix)
```

```python
correlation_pairs = correlation_matrix.where(
    ~correlation_matrix.isna()
).stack().reset_index()

correlation_pairs.columns = ["Column 1", "Column 2", "Correlation"]

correlation_pairs = correlation_pairs[
    correlation_pairs["Column 1"] < correlation_pairs["Column 2"]
]

correlation_pairs["Absolute Correlation"] = correlation_pairs["Correlation"].abs()

print(correlation_pairs.sort_values(
    "Absolute Correlation", ascending=False
).head(3))
```
